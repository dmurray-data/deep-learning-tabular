```python
# combine date and time into one feature

import pandas as pd

X["datetime"] = pd.to_datetime(
    X["Date"] + " " + X["Time"],
    dayfirst=True
)
```

```python
print("Start:", X["datetime"].min())
print("End:", X["datetime"].max())
```

```python
# convert measurement columns to numeric

numeric_cols = [
    "Global_active_power",
    "Global_reactive_power",
    "Voltage",
    "Global_intensity",
    "Sub_metering_1",
    "Sub_metering_2",
    "Sub_metering_3"
]

for col in numeric_cols:
    X[col] = pd.to_numeric(
        X[col],
        errors="coerce" # malformed strings will be converted to NaN
    )
```

```python
X.dtypes
```

```python
X[numeric_cols].isna().sum()
```

```python
X = X.sort_values("datetime").reset_index(drop=True)
```

```python
print("Rows:", len(X))
print("Start:", X["datetime"].min())
print("End:", X["datetime"].max())
```

```python
print(
    "Duplicate timestamps:",
    X["datetime"].duplicated().sum()
)
```

```python
# add missing timestamps

full_range = pd.date_range(
    start=X["datetime"].min(),
    end=X["datetime"].max(),
    freq="1min"
)

missing_timestamps = full_range.difference(
    X["datetime"]
)

print(
    "Missing timestamps:",
    len(missing_timestamps)
)
```

```python
# check if missing values occur together

missing_rows = X[
    X[numeric_cols].isna().any(axis=1)
]

print(missing_rows.shape)
missing_rows.head(20)
```

```python
# do the missing values occur in blocks?

missing_all = X[numeric_cols].isna().all(axis=1)

groups = (
    missing_all.ne(missing_all.shift())
    .cumsum()
)

missing_blocks = (
    X.loc[missing_all]
    .groupby(groups[missing_all])
    .agg(
        start=("datetime", "min"),
        end=("datetime", "max"),
        rows=("datetime", "size")
    )
)

missing_blocks.sort_values(
    "rows",
    ascending=False
).head(20)
```

```python
# considering the large blocks of missing values
# Preserve the missing periods, 
# construct the forecasting target and lag features while the full timeline is intact, 
# and then remove rows that cannot produce a valid training example.
```

```python
# full chronological timeline
#         ↓
# preserve NaNs/gaps
#         ↓
# create time features
#         ↓
# create lag features
#         ↓
# create 60-minute-ahead target
#         ↓
# identify valid examples
#         ↓
# drop unusable rows
```

```python
# final check for missing or dupe timestamps

full_range = pd.date_range(
    start=X["datetime"].min(),
    end=X["datetime"].max(),
    freq="1min"
)

missing_timestamps = full_range.difference(
    X["datetime"]
)

print("Expected timestamps:", len(full_range))
print("Actual rows:", len(X))
print("Missing timestamps:", len(missing_timestamps))
print("Duplicate timestamps:", X["datetime"].duplicated().sum())
```

### Missing-data finding

All seven electrical measurement variables contain 25,979 missing
observations, and the missing values occur together across measurement
columns.

The missingness is concentrated in contiguous outages rather than
isolated cells. Several gaps span multiple days; the six largest blocks
account for roughly 95% of all missing measurements.

Because long-gap interpolation would fabricate substantial amounts of
electrical-consumption data, missing measurements will not be
interpolated. The complete timestamp sequence will instead be preserved
while lag variables and the forecasting target are constructed. Rows
without sufficient historical predictors or a valid future target will
then be excluded from model training.

```python
# save to parquet

from pathlib import Path

processed_dir = Path("../data/processed")
processed_dir.mkdir(parents=True, exist_ok=True)

output_path = processed_dir / "household_power_validated.parquet"

X.to_parquet(
    output_path,
    index=False
)

print("Saved:", output_path)
print("Shape:", X.shape)
```
