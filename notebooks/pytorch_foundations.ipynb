# The Shape Rule

```python
X = torch.randn(512, 30) # 512 observations, 30 input features
W = torch.randn(30, 64) # layers weight matrix with 64 neurons

output = X @ W # Matrix multiplication gives 512 observations and 64 learned features

print("X shape:", X.shape)
print("W shape:", W.shape)
print("Output shape:", output.shape)
```

```python
# pytorch packages (same thing as above)
layer = torch.nn.Linear(30,64)

output = layer(X)

output.shape
```

```python
torch.nn.Linear(64,32) # another layer
```

```python
torch.nn.Linear(32,1) # final regression layer
```

```python
#    [512, 30]
#        ↓
#    30 real input features

#    [512, 64]
#        ↓
#    64 learned representations

#    [512, 32]
#        ↓
#    32 deeper learned representations

#    [512, 1]
#        ↓
#    1 prediction per observation
```

# Moving tensors from CPU to GPU

```python
import torch

device = torch.device("cuda" if torch.cuda.is_available() else "cpu") # checking for cuda device (gpu)

print("Using device:", device)
```

```python
x = torch.randn(512, 30) # creating a CPU tensor

print("x device:", x.device)
print("x shape:", x.shape)
```

```python
x_gpu = x.to(device) # moving tensor to "device" which was previosly defined as gpu cuda device

print("x_gpu device:", x_gpu.device)
print("x_gpu shape:", x_gpu.shape)
```

```python
W = torch.randn(30, 64, device=device) # W weights to gpu device

print("W device:", W.device)
print("W shape:", W.shape)
```

```python
output = x_gpu @ W # matrix multiplication on gpu!

print("Output shape:", output.shape)
print("Output device:", output.device)
```

```python
# reminder that all tensors in deep learning operations need to be on either CPU or GPU
# cannot have one tensor on CPU and one on GPU

# example
## cpu_tensor = torch.randn(512, 30)
## gpu_weights = torch.randn(30, 64, device=device)

## cpu_tensor @ gpu_weights # not on same chip, matrix multiplication attempt
```

# Autograd
### automatic differentiation 
#### tool for computing gradient of function with respect to its inputs

```python
x = torch.tensor(2.0, requires_grad=True)

y = x ** 2 # y = x^2  ->  dy/dx = 2x  ->  at x = 2  dy/dx should be 4

print("x:", x)
print("y:", y)
```

```python
y.backward() # the calculation (can only run once)
```

```python
print("Gradient:", x.grad) # expected value = 4
```

```python
# with more complex

x = torch.tensor(3.0, requires_grad=True)

y = 2 * x**2 + 5 * x + 1
```

```python
y.backward()

print(x.grad)
```

```python
# above is the foundation for backpropagation

# network version will look like

# input
#   ↓
# weights
#   ↓
# matrix multiplication
#   ↓
# activation
#   ↓
# more weights
#   ↓
# prediction
#   ↓
# loss              <-  loss.backward()
```
