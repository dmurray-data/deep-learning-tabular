```python
# tool import and test
import torch

print("PyTorch version:", torch.__version__)
print("CUDA available:", torch.cuda.is_available())
```

```python
scalar = torch.tensor(7) # a scalar tensor, one row, one column
vector = torch.tensor([1, 2, 3]) # a vector tensor, one row, three columns
matrix = torch.tensor([
    [1, 2, 3],
    [4, 5, 6]
]) # a matrix tensor

#row = observations, column = features

print("scalar:", scalar)
print("vector:", vector)
print("matrix:\n", matrix)
```

```python
print("scalar ndim:", scalar.ndim)
print("vector ndim:", vector.ndim)
print("matrix ndim:", matrix.ndim)

print("scalar shape:", scalar.shape)
print("vector shape:", vector.shape)
print("matrix shape:", matrix.shape)
```

```python
# tensor dtypes
int_tensor = torch.tensor([1, 2, 3])
float_tensor = torch.tensor([1.0, 2.0, 3.0])

print("int_tensor:", int_tensor)
print("dtype:", int_tensor.dtype)

print()

print("float_tensor:", float_tensor)
print("dtype:", float_tensor.dtype)
```

```python
x = torch.tensor(
    [1, 2, 3],
    dtype=torch.float32
)

print(x)
print(x.dtype)
```

```python
x = torch.tensor([1, 2, 3])

print(x)
print(x.dtype)

x_float = x.float()

print(x_float)
print(x_float.dtype)

# or x_float = x.to(torch.float32)
```

```python
# tensor indexing
x = torch.tensor([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

print(x)
print("shape:", x.shape)
```

# Retrieving Values

```python
# first row retrieval
x[0]
```

```python
# second row retrieval
x[1]
```

```python
# individual value retrieve
x[0, 0]
```

```python
x[1, 2]
```

```python
# slicing
x[:,0] # all rows, first column
```

```python
x[:,1] # all rows, second column
```

```python
x[:,:] # all rows, all columns
```

## Basic Tensor Math

```python
a = torch.tensor([1.0, 2.0, 3.0])
b = torch.tensor([4.0, 5.0, 6.0])
```

```python
# addition
a + b
```

```python
# subtraction
a - b
```

```python
# element wise multiplication
a * b
```

```python
# element wise division
a / b
```

# matrix multiplication

```python
A = torch.tensor([
    [1.0, 2.0],
    [3.0, 4.0]
])

B = torch.tensor([
    [5.0, 6.0],
    [7.0, 8.0]
])
```

```python
A @ B # matrix multiplication
```

```python
torch.matmul(A, B) # also matrix multiplication
```
