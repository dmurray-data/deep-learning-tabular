# Phase 2 Model Comparison

## Goal

Evaluate whether the expanded Phase 2 temporal feature set improves
60-minute-ahead household power forecasting performance.

Phase 1 showed that Histogram Gradient Boosting was the strongest validation model,
while the PyTorch neural network performed competitively but slightly worse.

Phase 2 adds new temporal features describing:

- recent power changes
- short-term volatility
- recent maximum power
- recent minimum power
- recent power range

The purpose of this notebook is to determine whether these additional features
provide measurable forecasting improvement.

## Phase 1 Benchmarks

### Histogram Gradient Boosting

- RMSE: `0.7885`
- MAE: `0.5264`
- R²: `0.434`

### Baseline PyTorch Neural Network

- RMSE: approximately `0.7983`
- Performed approximately `1.24%` worse than HGB

Histogram Gradient Boosting therefore remains the current model to beat.

## Phase 2 Experimental Question

Does the expanded 31-feature temporal dataset improve validation performance
compared with the original 22-feature Phase 1 dataset?

## Plan

1. Load the Phase 2 feature dataset.
2. Recreate the same chronological 70/15/15 train, validation, and test split.
3. Verify that the test set remains untouched during model development.
4. Train Histogram Gradient Boosting using the 31-feature Phase 2 dataset.
5. Evaluate HGB on the validation set using:
   - MAE
   - RMSE
   - R²
6. Compare Phase 2 HGB performance against the Phase 1 HGB benchmark.
7. If the expanded features provide useful improvement, retrain the baseline
   PyTorch MLP using the same Phase 2 features.
8. Compare the Phase 2 neural network against:
   - Phase 1 neural network
   - Phase 1 HGB
   - Phase 2 HGB
9. Select the strongest model using validation evidence.
10. Evaluate the selected final model or models on the untouched test set.

## Experimental Rules

- Use the same chronological split proportions as Phase 1.
- Do not use random train/test splitting.
- Do not use the test set for tuning or model selection.
- Keep the HGB hyperparameters initially unchanged so the effect of the new
  feature set can be measured cleanly.
- Compare models using the same validation period and evaluation metrics.

## Success Criteria

The Phase 2 features will be considered useful if they produce a meaningful
reduction in validation RMSE compared with the Phase 1 benchmarks.

The first benchmark to beat is:

`Phase 1 HGB RMSE = 0.7885`
